## <u>Bài 1</u>

Green M&M Candies liên quan đến Dataset 18 trong file excel. Tìm tỉ lệ mẫu của M&M có màu xanh lá cây (green). Sử dụng kết quả để xây dựng 1 ước lượng khoảng tin cậy 95% của % quần thể M&M có màu xanh.

```python
import numpy as np
import pandas as pd
from scipy import stats  # Thêm dòng này để sửa lỗi NameError
```

```python
# 2. Tính tổng số viên M&M (n) và số viên màu xanh (x)
n_total = df.count().sum()          # Tổng số ô khác NaN ở tất cả các cột
x_green = df['Green'].count()       # Số ô khác NaN ở cột Green
# 3. Tỉ lệ mẫu (p_hat)
p_hat = x_green / n_total

# 4. Tính khoảng tin cậy 95%
confidence_level = 0.95
alpha = 1 - confidence_level
z_critical = stats.norm.ppf(1 - alpha / 2)

# Sai số chuẩn (Standard Error) và Sai số ước lượng (Margin of Error)
se = np.sqrt(p_hat * (1 - p_hat) / n_total)
margin_of_error = z_critical * se

ci_lower = p_hat - margin_of_error
ci_upper = p_hat + margin_of_error

# 5. In kết quả
print("=== KẾT QUẢ BÀI 1 ===")
print(f"- Tổng số viên M&M (n): {n_total}")
print(f"- Số viên M&M màu xanh (x): {x_green}")
print(f"- Tỉ lệ mẫu (p_hat): {p_hat:.4f} ({p_hat * 100:.2f}%)")
print(f"- Sai số ước lượng (Margin of Error): ±{margin_of_error * 100:.2f}%")
print(f"- Khoảng tin cậy 95% cho % quần thể: [{ci_lower * 100:.2f}%, {ci_upper * 100:.2f}%]")
```

## <u>Bài 2</u>

Freshman Weight Gain liên quan đến Dataset 3 trong file excel

a. Dựa vào kết quả của mẫu, tìm ước lượng điểm tốt nhất của tỉ lệ phần trăm các sinh viên cao đẳng tăng cân trong năm thứ 1.

b. Xây dựng ước lượng khoảng tin cậy 95% về tỉ lệ phần trăm các sinh viên cao đẳng tăng cân trong năm thứ 1.


```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Đọc dữ liệu từ file Excel Dataset 3

df = pd.read_excel(r'D:\QuocAnh\TKMTUD\TKMT_lab_06\TKMT_lab_06\data\03_FRESH15.xls')

# 2. Xác định điều kiện tăng cân
# Trong Dataset 3, sự thay đổi cân nặng thường được tính bằng (Cân nặng Tháng 4 - Cân nặng Tháng 9)
# Sinh viên tăng cân có chênh lệch cân nặng > 0 (hoặc WT_APR > WT_SEP)
if 'Weight_Change' in df.columns:
    gained_weight = df['Weight_Change'] > 0
elif 'WT_APR' in df.columns and 'WT_SEP' in df.columns:
    gained_weight = df['WT_APR'] > df['WT_SEP']
else:
    # Nếu file chỉ có 1 cột chênh lệch cân nặng
    gained_weight = df.iloc[:, -1] > 0

# Tổng số sinh viên (n) và số sinh viên tăng cân (x)
n = len(df.dropna())
x = gained_weight.sum()

# === Câu a: Ước lượng điểm tốt nhất ===
p_hat = x / n

print("=== BÀI 2: FRESHMAN WEIGHT GAIN ===")
print(f"a. Ước lượng điểm tốt nhất của tỉ lệ tăng cân (p_hat): {p_hat:.4f} ({p_hat * 100:.2f}%)")

# === Câu b: Khoảng tin cậy 95% ===
confidence = 0.95
alpha = 1 - confidence
z_critical = stats.norm.ppf(1 - alpha / 2)  # z_0.025 ≈ 1.96

# Sai số chuẩn và Sai số ước lượng
se = np.sqrt(p_hat * (1 - p_hat) / n)
margin_of_error = z_critical * se

# Cận dưới và Cận trên
ci_lower = p_hat - margin_of_error
ci_upper = p_hat + margin_of_error

print(f"\nb. Ước lượng khoảng tin cậy 95%:")
print(f"   - Kích thước mẫu (n): {n}")
print(f"   - Số sinh viên tăng cân (x): {x}")
print(f"   - Giá trị z_critical (95%): {z_critical:.4f}")
print(f"   - Sai số ước lượng (E): ±{margin_of_error * 100:.2f}%")
print(f"   - Khoảng tin cậy 95%: [{ci_lower * 100:.2f}%, {ci_upper * 100:.2f}%]")
```

## <u>Bài 3</u>

Lượng mưa ở Boston: liên quan đến Dataset 14 trong file excel, và quan tâm đến các ngày với các giá trị lượng mưa khác nhau từ 0 đến các ngày có mưa có giá trị lượng mưa lớn hơn 0. Xây dựng ước lượng khoảng tin cậy 95% cho tỉ lệ mưa trong các ngày Thứ Tư và xây dựng ước lượng khoảng tin cậy 95% cho tỉ lệ mưa trong các ngày Chủ Nhật.

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Đọc dữ liệu từ file Excel Dataset 14
df = pd.read_excel(r'D:\QuocAnh\TKMTUD\TKMT_lab_06\TKMT_lab_06\data\14_BOSTRAIN.xls')

# Hàm tính khoảng tin cậy 95% cho tỉ lệ mưa
def calc_rain_ci_95(precip_data, day_name=""):
    # Loại bỏ giá trị khuyết (NaN)
    clean_data = precip_data.dropna()
    
    # Tổng số ngày (n) và số ngày có mưa > 0 (x)
    n = len(clean_data)
    x = (clean_data > 0).sum()
    
    # Tỉ lệ mẫu (p_hat)
    p_hat = x / n
    
    # Mức tin cậy 95% -> z_critical = 1.96
    confidence = 0.95
    alpha = 1 - confidence
    z_critical = stats.norm.ppf(1 - alpha / 2)
    
    # Sai số ước lượng (Margin of Error)
    se = np.sqrt(p_hat * (1 - p_hat) / n)
    margin_of_error = z_critical * se
    
    ci_lower = p_hat - margin_of_error
    ci_upper = p_hat + margin_of_error
    
    print(f"=== {day_name.upper()} ===")
    print(f"- Tổng số ngày quan sát (n): {n}")
    print(f"- Số ngày có mưa (x > 0): {x}")
    print(f"- Tỉ lệ mẫu (p_hat): {p_hat:.4f} ({p_hat * 100:.2f}%)")
    print(f"- Sai số ước lượng (E): ±{margin_of_error * 100:.2f}%")
    print(f"- Khoảng tin cậy 95%: [{ci_lower * 100:.2f}%, {ci_upper * 100:.2f}%]\n")
    
    return p_hat, (ci_lower, ci_upper)

# 2. Kiểm tra cấu trúc file và tính toán
print("=== BÀI 3: ƯỚC LƯỢNG TỈ LỆ MƯA Ở BOSTON (DATASET 14) ===\n")

# Dạng 1: Các ngày được chia thành từng cột riêng (WED, SUN)
if 'WED' in df.columns or 'Wednesday' in df.columns:
    wed_col = 'WED' if 'WED' in df.columns else 'Wednesday'
    sun_col = 'SUN' if 'SUN' in df.columns else 'Sunday'
    
    calc_rain_ci_95(df[wed_col], "Thứ Tư")
    calc_rain_ci_95(df[sun_col], "Chủ Nhật")

# Dạng 2: Cấu trúc bảng gồm cột Day/Ngày và Cột Precipitation/Lượng mưa
elif 'Day' in df.columns or 'Ngay' in df.columns:
    day_col = 'Day' if 'Day' in df.columns else 'Ngay'
    precip_col = [c for c in df.columns if c != day_col][0]
    
    wed_data = df[df[day_col].astype(str).str.contains('Wed', case=False)][precip_col]
    sun_data = df[df[day_col].astype(str).str.contains('Sun', case=False)][precip_col]
    
    calc_rain_ci_95(wed_data, "Thứ Tư")
    calc_rain_ci_95(sun_data, "Chủ Nhật")
```

## <u>Bài 4</u>

Tổng số tiền phim: liên quan đến Dataset 9 trong file excel. Xây dựng ước tính khoảng thời gian tin cậy 95% của tổng số tiền trung bình cho quần thể của tất cả các phim. Giả định rằng độ lệch chuẩn của quần thể được biết là 100 triệu đô la. 

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Đọc dữ liệu từ file Excel Dataset 9
df = pd.read_excel(r'D:\QuocAnh\TKMTUD\TKMT_lab_06\TKMT_lab_06\data\09_MOVIES.xls')

# 2. Tìm cột chứa doanh thu phim
possible_cols = ['Gross', 'Gross_Revenue', 'Gross ($M)', 'Gross_M', 'Revenue']
target_col = None
for col in df.columns:
    if any(p.lower() in col.lower() for p in possible_cols):
        target_col = col
        break

if target_col is None:
    # Nếu không tìm thấy tên cột trùng khớp, lấy cột định dạng số đầu tiên
    target_col = df.select_dtypes(include=[np.number]).columns[0]

# Lấy dữ liệu doanh thu và làm sạch (loại bỏ NaN)
gross_data = df[target_col].dropna()

# 3. Tính các đại lượng thống kê mẫu
n = len(gross_data)             # Số lượng phim trong mẫu
x_bar = gross_data.mean()        # Trung bình mẫu (triệu USD)
sigma = 100.0                    # Độ lệch chuẩn quần thể đã biết (triệu USD)

# 4. Tính khoảng tin cậy 95%
confidence = 0.95
alpha = 1 - confidence
z_critical = stats.norm.ppf(1 - alpha / 2)  # z_0.025 ≈ 1.9600

# Sai số chuẩn (Standard Error) và Sai số ước lượng (Margin of Error)
standard_error = sigma / np.sqrt(n)
margin_of_error = z_critical * standard_error

ci_lower = x_bar - margin_of_error
ci_upper = x_bar + margin_of_error

# Hoặc tính trực tiếp bằng hàm của scipy.stats:
# ci_lower, ci_upper = stats.norm.interval(confidence, loc=x_bar, scale=standard_error)

# 5. In kết quả chi tiết
print("=== BÀI 4: TỔNG SỐ TIỀN PHIM (DATASET 9) ===")
print(f"- Cột dữ liệu phân tích: '{target_col}'")
print(f"- Số lượng phim trong mẫu (n): {n}")
print(f"- Doanh thu trung bình mẫu (x_bar): {x_bar:.2f} triệu USD")
print(f"- Độ lệch chuẩn quần thể (sigma): {sigma:.2f} triệu USD")
print(f"- Giá trị z_critical (95%): {z_critical:.4f}")
print(f"- Sai số ước lượng (E): ±{margin_of_error:.2f} triệu USD")
print(f"- Khoảng tin cậy 95% cho doanh thu trung bình quần thể (mu):")
print(f"  [{ci_lower:.2f} triệu USD,  {ci_upper:.2f} triệu USD]")
```

## <u>Bài 5</u>

Điểm đánh giá tín dụng FICO: liên quan đến Dataset 24 trong file excel. Xây dựng ước lượng khoảng tin cậy 99% của điểm FICO trung bình cho quần thể. Giả sử độ lệch chuẩn của quần thể là 92,2.

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Đọc dữ liệu từ file Excel Dataset 24
df = pd.read_excel(r'D:\QuocAnh\TKMTUD\TKMT_lab_06\TKMT_lab_06\data\24_FICO.xls')

# 2. Tìm cột chứa điểm FICO
possible_cols = ['FICO', 'FICO_Score', 'Score', 'Credit_Score']
target_col = None
for col in df.columns:
    if any(p.lower() in str(col).lower() for p in possible_cols):  # An toàn cho mọi kiểu dữ liệu
        target_col = col
        break

if target_col is None:
    # Nếu không tìm thấy tên cột trùng khớp, lấy cột định dạng số đầu tiên
    target_col = df.select_dtypes(include=[np.number]).columns[0]

# Lấy dữ liệu điểm FICO và loại bỏ giá trị khuyết (NaN)
fico_data = df[target_col].dropna()

# 3. Tính các đại lượng thống kê
n = len(fico_data)               # Kích thước mẫu
x_bar = fico_data.mean()          # Trung bình mẫu điểm FICO
sigma = 92.2                      # Độ lệch chuẩn quần thể đã biết

# 4. Tính khoảng tin cậy 99%
confidence = 0.99
alpha = 1 - confidence
z_critical = stats.norm.ppf(1 - alpha / 2)  # z_0.005 ≈ 2.5758

# Sai số chuẩn (Standard Error) và Sai số ước lượng (Margin of Error)
standard_error = sigma / np.sqrt(n)
margin_of_error = z_critical * standard_error

ci_lower = x_bar - margin_of_error
ci_upper = x_bar + margin_of_error

# Hoặc tính trực tiếp bằng hàm scipy.stats.norm.interval:
# ci_lower, ci_upper = stats.norm.interval(confidence, loc=x_bar, scale=standard_error)

# 5. In kết quả chi tiết
print("=== BÀI 5: ĐIỂM ĐÁNH GIÁ TÍN DỤNG FICO (DATASET 24) ===")
print(f"- Cột dữ liệu phân tích: '{target_col}'")
print(f"- Kích thước mẫu (n): {n}")
print(f"- Điểm FICO trung bình mẫu (x_bar): {x_bar:.2f}")
print(f"- Độ lệch chuẩn quần thể (sigma): {sigma:.2f}")
print(f"- Giá trị z_critical (99%): {z_critical:.4f}")
print(f"- Sai số ước lượng (E): ±{margin_of_error:.2f}")
print(f"- Khoảng tin cậy 99% cho điểm FICO trung bình quần thể (mu):")
print(f"  [{ci_lower:.2f}, {ci_upper:.2f}]")
```
