# CÂU 5. Phân tích theo loại thực phẩm – 1,5 điểm

Tính giá trị trung bình của `consumption` và `co2_emission` theo từng `food_category`.

Tạo DataFrame gồm:
- `food_category`
- `avg_consumption`
- `avg_co2_emission`

Sắp xếp theo `avg_co2_emission` giảm dần và xác định 3 loại thực phẩm có lượng phát thải CO₂ trung bình cao nhất.

```python
# Tính trung bình theo cột food_category
kq=df.groupby('food_category')[['consumption','co2_emission']].mean()
# Đưa kq này thành cột ??
kq=kq.reset_index()
# đổi tên:
kq=kq.rename(columns={'consumption':'avg_consumption',
                    'co2_emission':'avg_co2_emission' })
# Sắp xếp giảm dần
kq=kq.sort_values('avg_co2_emission',ascending=False)
# Xuất 3 loại thực phẩm có CO2 cao nhất
top3=kq.head(3)
print(kq)
print("\n")
print(top3)
```

## Câu 6. Biểu đồ cột bằng Matplotlib – 1.5 điểm

Sử dụng **Matplotlib** vẽ biểu đồ cột thể hiện **mức tiêu thụ trung bình của từng loại thực phẩm**.

Yêu cầu:
- Trục X: `food_category`.
- Trục Y: mức tiêu thụ trung bình.
- Có tiêu đề: `Mức tiêu thụ trung bình theo loại thực phẩm`.
- Có tên hai trục.

```python
import matplotlib.pyplot as plt
x=kq['food_category']
y=kq['avg_consumption']
plt.bar(x,y)
plt.title('Mức tiêu thụ trung bình theo loại thực phẩm')
plt.xlabel('Food_category')
plt.ylabel('Mức tiêu thụ trung bình')
plt.show()
```

# CÂU 7. Phát hiện Outlier bằng IQR – 1,5 điểm

Đối với `co2_emission`, tính Q1, Q3, IQR, Lower Bound và Upper Bound.

Sau đó xác định số lượng outlier trên **toàn bộ dữ liệu** theo quy tắc 1.5 × IQR.

`IQR = Q3 - Q1`

`Lower = Q1 - 1.5 × IQR`

`Upper = Q3 + 1.5 × IQR`

**Lưu ý:** Không tính riêng theo từng `food_category`.

```python
Q1=df['co2_emission'].quantile(0.25)
Q3=df['co2_emission'].quantile(0.75)
IQR=Q3-Q1
print("IQR =",IQR)
Lower =Q1-1.5*IQR
Upper =Q3+1.5*IQR
print("Lower =",Lower)
print("Upper =",Upper)
outliers=df[
    (df['co2_emission']<Lower) | (df['co2_emission']>Upper)
]
print("Số lượng outlier = ",len(outliers))
```
